Read the Estuary_fish.csv file
• save samples with:
• management == Protected, Shore == Healthy and length > 2.0
• to a new file

```python
import pandas as pd

# Read the CSV file
df = pd.read_csv("Estuary_fish.csv")
df
```

```python
# Filter samples based on the given conditions
filtered_df = df[(df['Management'] == 'Protected') & 
                 (df['Shore'] == 'Healthy') & 
                 (df['Length'] > 2.0)]
filtered_df

# The system is sensitive to uppercase letters
```

```python
# Save the filtered data to a new CSV file
filtered_file_path = "Filtered_Estuary_fish.csv"
filtered_df.to_csv(filtered_file_path, index=False)

print(f"Filtered data saved to '{filtered_file_path}'.")
```
