```python
'''
List Comprehension
List comprehension offers a shorter syntax when you want to create a new list based
on the values of an existing list.
newlist = [expression for item in iterable if condition == True]
example:
newlist = [x for x in fruits if x != "apple"]
newlist = [x for x in range(10)]
list comprehensions are generally faster than traditional for loops in Python.
'''
longstring = "list one oie haha tuduuuu"
result = [item for item in longstring.split() if len(item) >=5]
print(result)
```

```python
'''
Q1: Write a Python program that performs the following tasks:
• Given a list of strings, filter out all strings that are shorter than 5 characters.
• Convert the remaining strings to uppercase.
• Sort the resulting list alphabetically.
• Print the transformed list.
• Try to use as fewer lines as possible.
'''
longstring = "list one oie haha tuduuuu"
result = sorted(s.upper() for s in longstring.split() if len(s) >= 5)
print(result)
```

```python
strings = ["apple", "bat", "cat", "elephant", "dog", "zebra"]
result = sorted(s.upper() for s in strings if len(s) >= 5)
print(result)
```

```python
# Q2: Write a python program to implement the Selection Sort algorithm.

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_ind = i
        for j in range(i+1, n):
            if arr[j] < arr[min_ind]:
                min_ind = j
        arr[i], arr[min_ind] = arr[min_ind], arr[i]

numbers = [64, 25, 12, 22, 11]
selection_sort(numbers)
print(numbers)
```
