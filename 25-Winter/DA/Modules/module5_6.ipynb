### DataFrame 
dataframe has both a row and column index; it can be thought of as a dict of Series all sharing the same index.

```python
data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada', 'Nevada'],
            'year': [2000, 2001, 2002, 2001, 2002, 2003],
            'pop': [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}
df = pd.DataFrame(data)
```

```python
df
```

```python
# df.head()

pd.DataFrame(data, columns=['year', 'state', 'pop']) #column order
```

```python
# column
df['state']
df.state # only works when the column name is a valid Python variable name

```

```python
# row
df.index = ['zero',"one",'two','three','four','five']
# df
df.loc['three'] 
```

```python
# When you are assigning lists or arrays to a column, the value’s length must match the length of the DataFrame. 
# If you assign a Series, its labels will be realigned exactly to the DataFrame’s index, inserting missing values in any holes:
frame2 = pd.DataFrame(data, columns=['year', 'state', 'pop', 'debt'],
  index=['one', 'two', 'three', 'four',
      'five', 'six'])
val = pd.Series([-1.2, -1.5, -1.7], index=['two', 'four', 'five'])
frame2['debt'] = val
```

```python
frame2
```

```python
# new column
frame2['eastern'] = frame2.state == 'Ohio'
# remove column
del frame2['eastern']
```

```python
# another way to create dataframe
pop = {'Nevada': {2001: 2.4, 2002: 2.9}, 'Ohio': {2000: 1.5, 2001: 1.7, 2002: 3.6}}
df = pd.DataFrame(pop)
df
```

```python
# As with Series, the values attribute returns the data contained in the DataFrame as a two-dimensional ndarray:
df.values
```

```python
df.index = ["1","2","3"]
df.columns = ["one","two"]
df
```

```python
# reindexing
# create a new object with the data conformed to a new index. 
obj = pd.Series([4.5, 7.2, -5.3, 3.6], index=['d', 'b', 'a', 'c'])
obj2 = obj.reindex(['a', 'b', 'c', 'd', 'e'])
obj2
```

```python
frame = pd.DataFrame(np.arange(9).reshape((3, 3)),
                     index=['a', 'c', 'd'],
                     columns=['Ohio', 'Texas', 'California'])

frame2 = frame.reindex(['a', 'b', 'c', 'd'])
frame3 = frame.reindex(columns=['Texas', 'Utah', 'California'])
```

```python
frame3
```

```python
#drop
obj = pd.Series(np.arange(5.), index=['a', 'b', 'c', 'd', 'e'])
obj = obj.drop(['d', 'c'])

obj

# Many functions, like drop, which modify the size or shape of a Series or DataFrame, can manipulate an object in-place without returning a new object
```

```python
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=['Ohio', 'Colorado', 'Utah', 'New York'],
                    columns=['one', 'two', 'three', 'four'])

print(data)
data = data.drop(['Colorado', 'Ohio'])
print(data)
data=  data.drop(['two', 'four'], axis='columns') # axis=1
print(data)
```

```python
# Indexing into a DataFrame is for retrieving one or more columns either with a single value or sequence:
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=['Ohio', 'Colorado', 'Utah', 'New York'],
                    columns=['one', 'two', 'three', 'four'])

# data[:2]
print(data)
print(data[data['three'] > 5])
```

```python
# applying a function on one-dimensional arrays to each column or row
frame = pd.DataFrame(np.random.randn(4, 3), columns=list('bde'),
                     index=['Utah', 'Ohio', 'Texas', 'Oregon'])
f = lambda x: x.max() - x.min()
frame.apply(f)
# f computes the difference between the maximum and mininum of a Series, 
# is invoked once on each column in frame.
```

```python
# Element-wise Python functions
format = lambda x: '%.2f' % x
frame.applymap(format)
```
