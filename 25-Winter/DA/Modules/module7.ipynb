### Data cleaning and preparation

```python
from numpy import nan as NA
import pandas as pd
import numpy as np
```

![alt text](na.png "Title")

```python
data = pd.Series([1, NA, 3.5, NA, 7])
data.dropna()
```

```python
data = pd.DataFrame([[1., 6.5, 3.], [1., NA, NA],
                     [NA, NA, NA], [NA, 6.5, 3.]])
print(data)

cleaned = data.dropna() # default for rows

print("\n")
print(cleaned)
```

```python
data.dropna(axis=1, how='all') # drop columns with all NA values
```

```python
# Filling In Missing Data
df = pd.DataFrame(np.random.randn(7, 3))
df.iloc[:4, 1] = NA
df.iloc[:2, 2] = NA

print(df)

print(df.fillna(0))

print(df.fillna({1: 0.5, 2: 0})) # fill column 1 with 0.5 and column 2 with 0
```

```python
df.ffill()
```

```python
df.bfill()
```

```python
print(data.fillna(data.mean())) # fill with mean of data
```

```python
# duplicates

data = pd.DataFrame({'k1': ['one', 'two'] * 3 + ['two'],
                     'k2': [1, 1, 2, 3, 3, 4, 4]})
print(data)

data.duplicated()

```

```python
data.drop_duplicates()
```

```python
# Transforming Data Using a Function or Mapping

data = pd.DataFrame({'food': ['bacon', 'pulled pork', 'bacon',
                              'Pastrami', 'corned beef', 'Bacon',
                              'pastrami', 'honey ham', 'nova lox'],
                              'ounces': [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
print(data)
meat_to_animal = {
      'bacon': 'pig',
      'pulled pork': 'pig',
      'pastrami': 'cow',
      'corned beef': 'cow',
      'honey ham': 'pig',
      'nova lox': 'salmon'
}

lowercased = data['food'].str.lower()

data['animal'] = lowercased.map(meat_to_animal)
print(data)
```

```python
data = pd.Series([1., -999., 2., -999., -1000., 3.])
data.replace(-999, np.nan)
```

```python
# filtering out outliers

data = pd.DataFrame(np.random.randn(1000, 4))

data.describe()
```

```python
data[np.abs(data) > 3] = np.sign(data) * 3
print(data.describe())
```
