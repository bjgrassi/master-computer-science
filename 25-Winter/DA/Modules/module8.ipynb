### Data Wrangling: Join, Combine, and Reshape

```python
import pandas as pd
import numpy as np
```

```python
# combine Series into DataFrame

df1 = pd.DataFrame({'key': ['b', 'b', 'a', 'c', 'a', 'a', 'b'],
                    'data1': range(7)})
df2 = pd.DataFrame({'key': ['a', 'b', 'd'],
                    'data2': range(3)})

print(df1)
print(df2)
```

```python
pd.merge(df1, df2) # merge by key (common columns), only common keys are merged
```

```python
# If the column names are different in each object, you can specify them separately
df3 = pd.DataFrame({'lkey': ['b', 'b', 'a', 'c', 'a', 'a', 'b'],
                    'data1': range(7)})
df4 = pd.DataFrame({'rkey': ['a', 'b', 'd'],
                    'data2': range(3)})
```

```python
print(df3)
print(df4)
```

```python
pd.merge(df3, df4, left_on='lkey', right_on='rkey') # By default merge does an 'inner' join; other options are 'left', 'right', 'outer'
```

```python
pd.merge(df1, df2, how='outer') # The outer join takes the union of the keys
```

```python
# Pivoting between "Long" to "Wide" Format
data = pd.read_csv('./macrodata.csv')
```

```python
data.head()
```

```python
periods = pd.PeriodIndex.from_fields(year=data.year, quarter=data.quarter, freq='Q') # create a period index
periods.name = 'date'
columns = pd.Index(['realgdp', 'infl', 'unemp'], name='item')
data = data.reindex(columns=columns)
data.index = periods.to_timestamp('D', 'end') # convert to timestamp, 'D' means day, 'end' means the end of the day
df_long = data.stack().reset_index().rename(columns={0: 'value'})

# print(df_long.head())
print(df_long)
```

```python
# Data is frequently stored this way in relational databases like MySQL
pivoted = df_long.pivot(index='date', columns='item', values='value')
print(pivoted.head())
```

```python
df = pd.DataFrame({'key': ['foo', 'bar', 'baz'],
                   'A': [1, 2, 3],
                   'B': [4, 5, 6],
                   'C': [7, 8, 9]})

df
```

```python
melted = pd.melt(df, ['key']) # melt is the inverse of pivot
melted
```

```python
reshaped = melted.pivot(index='key', columns='variable', values='value')
print(reshaped)
```

```python
reshaped.reset_index() # Since the pivot result has a hierarchical index, using reset_index will turn the data back into a regular DataFrame
```

### Q1
Solve the following questions based on the given dataframe

1. Standardize the 'product' column by converting all values to lowercase
2. Replace the -999 value in the 'quantity' column with NaN
3. Remove any duplicate rows based on the 'product' and 'price' columns
4. Calculate the average price per product type

```python
data = pd.DataFrame({
    'product': ['Apple', 'Banana', 'Orange', 'apple', 'BANANA', 'orange', 'Apple'],
    'price': [1.5, 0.5, 0.8, 1.2, 0.5, 0.8, 1.2],
    'quantity': [10, 20, 15, 5, -999, 18, 8],
    'store': ['A', 'B', 'A', 'C', 'B', 'C', 'A']
})
data
```

```python
# 1. 
data['product'] = data['product'].str.lower()

# 2. a
data['quantity'] = data['quantity'].replace(-999, np.nan)

# 3. 
data = data.drop_duplicates(subset=['product', 'price'])

print(data)
```

```python
# 4.
average_price = data.groupby('product')['price'].mean()
print(average_price)
```
