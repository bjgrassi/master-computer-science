### Q1: (3 points)
Write a Python program to create 100 random integers and print out all the prime numbers.

```python
random_arr = np.random.randint(1, 1001, 100)
print('Random array:\n', random_arr)

def is_prime(num):
    # all prime numbers are greater than 1
    if num > 1:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                return False
        return True


prime_numbers_filtered = [int(num) for num in random_arr if is_prime(num)]
print("\nPrime numbers:", prime_numbers_filtered)
```

###  Q2: (3 points)
Create a NumPy array of shape (4, 3) with random integers between 1 and 50.

Write code to:
- Find the maximum value in the array.
- Calculate the mean of each row.
- Replace all values greater than 25 with 0.

```python
arr = np.random.randint(1, 51, (4,3))
print(arr)

print("\nMaximum value in the array:", np.max(arr))

row_means = np.mean(arr, axis=1)
print("\nMean of each row:\n", row_means)

arr[arr > 25] = 0
print("\n Array with values > 25 replaced with 0:\n", arr)
```

### Q3: (4 points)
Write a Python program to perform the following tasks:

1. Generate a 5×6 matrix, X, with random values sampled from a normal distribution (mean = 0, standard deviation = 1).
2. Perform the following calculation for each element in X:
Y = 1 / 1+e^x
3. Save the resulting matrix Y into a new file named result.txt, ensuring the matrix is formatted for readability.

```python
X = np.random.normal(0, 1, (5, 6))
print("Matrix X:\n", X)

Y = 1 / (1 + np.exp(X)) 
print("\nCalculation in Y:\n", Y)

np.savetxt("result.txt", Y, fmt="%.8f")
print("\nFile created!")
```
